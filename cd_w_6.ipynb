```python

!apt-get update -qq
!apt-get install -y flex gcc -qq

lex_code = r'''
%{
#include <stdio.h>
int paren = 0, square = 0, question = 0;
%}

%%
"("     { printf("TOKEN: (\n"); paren++; }
")"     { printf("TOKEN: )\n"); paren--; }
"["     { printf("TOKEN: [\n"); square++; }
"]"     { printf("TOKEN: ]\n"); square--; }
"?"     { printf("TOKEN: ?\n"); question++; }
[ \t\n]+ ;
.       ;
%%

int yywrap(void) { return 1; }

int main(void)
{
    printf("Enter expression:\n");
    yylex();

    if (paren == 0 && square == 0)
        printf("Parentheses are Balanced\n");
    else
        printf("Parentheses are Unbalanced\n");

    if (square == 0)
        printf("Square Brackets are Balanced\n");
    else
        printf("Square Brackets are Unbalanced\n");

    printf("Total '?' operators found: %d\n", question);
    return 0;
}
'''

with open("math.l", "w") as f:
    f.write(lex_code)

!flex math.l
!gcc lex.yy.c -o math
!echo 'a = (b + c) * arr[5]; x = (a > b) ? a : b;' | ./math
```
