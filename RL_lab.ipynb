```python

# PRACTICAL 10
# Bellman Optimality and Value Iteration
# Deterministic 2x2 Grid World

# Step 1: Define states, actions and parameters

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
epsilon = 0.001

# Step 2: Define transitions and rewards

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },
    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 10)
    }
}

# Step 3: Display MDP

print("GRID WORLD")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

# Step 4: Initialize values

V = {'S0': 0, 'S1': 0, 'G': 0}

print("\nInitial Values:", V)

# Step 5: Perform Value Iteration

iteration = 0

while True:
    iteration += 1
    new_V = V.copy()

    for state in ['S0', 'S1']:
        values = []

        for action in actions:
            next_state, reward = transitions[state][action]

            value = reward + gamma * V[next_state]
            values.append(value)

        new_V[state] = max(values)

    delta = max(abs(new_V[s] - V[s])
                for s in ['S0', 'S1'])

    V = new_V

    print(f"Iteration {iteration}: {V}, Delta={delta:.4f}")

    if delta < epsilon:
        break

# Step 6: Calculate Q values

print("\nOPTIMAL Q VALUES")

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]

        Q[state][action] = reward + gamma * V[next_state]

        print(state, action, "=", round(Q[state][action], 4))

# Step 7: Find Optimal Policy

print("\nOPTIMAL POLICY")

for state in ['S0', 'S1']:
    maximum = max(Q[state].values())

    best_actions = [
        action for action in actions
        if abs(Q[state][action] - maximum) < 1e-9
    ]

    print(state, ":", best_actions)

# Step 8: Verify Bellman Optimality

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in ['S0', 'S1']:
    print("V*(", state, ") =", round(V[state], 4))
    print("Max Q* =", round(max(Q[state].values()), 4))

    assert abs(V[state] - max(Q[state].values())) < 1e-6

print("\nValue Iteration Completed Successfully!")
```
