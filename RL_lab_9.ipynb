```python

# Practical 9 - Assignment 1
# Bellman Expectation Equations
# Stochastic Policy, Deterministic Environment

gamma = 0.9

# Solve the simultaneous Bellman equations algebraically.
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

V0 = 88.919 / (1 - 0.81 * 0.09)
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 89.9 + 0.09 * V0
Q_S1_Up = -1 + gamma * V0

print("ASSIGNMENT 1: ALGEBRAIC SOLUTION")
print("-" * 40)
print(f"V(S0)       = {V0:.2f}")
print(f"V(S1)       = {V1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right)= {Q_S0_Right:.2f}")
print(f"Q(S1, Right)= {Q_S1_Right:.2f}")
print(f"Q(S1, Up)   = {Q_S1_Up:.2f}")

# Verify the Bellman expectation equations
V0_check = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
V1_check = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("\nVerification:")
print(f"V(S0) verified: {abs(V0 - V0_check) < 1e-9}")
print(f"V(S1) verified: {abs(V1 - V1_check) < 1e-9}")
```

```python

# Practical 9 - Assignment 2
# Iterative Policy Evaluation
# Deterministic Policy, Stochastic Environment

gamma = 0.9
theta = 0.000001

# Initial state values
V = {
    "S0": 0.0,
    "S1": 0.0
}

iterations = 0

# Iterative policy evaluation
while True:
    old_V0 = V["S0"]
    old_V1 = V["S1"]

    # S0: Down
    # 0.1 probability -> Wall, reward -1
    # 0.9 probability -> S1, reward -1
    new_V0 = (
        0.1 * (-1)
        + 0.9 * (-1 + gamma * old_V1)
    )

    # S1: Right
    # 0.9 probability -> Goal, reward +100
    # 0.1 probability -> S0, reward -1
    new_V1 = (
        0.9 * 100
        + 0.1 * (-1 + gamma * old_V0)
    )

    # Update values simultaneously
    V["S0"] = new_V0
    V["S1"] = new_V1

    iterations += 1

    # Maximum change between iterations
    delta = max(
        abs(new_V0 - old_V0),
        abs(new_V1 - old_V1)
    )

    if delta < theta:
        break

# Calculate action values after convergence
Q_S0_Down = -1 + gamma * V["S1"]
Q_S0_Right = -1
Q_S1_Right = 0.9 * 100 + 0.1 * (
    -1 + gamma * V["S0"]
)
Q_S1_Up = -1 + gamma * V["S0"]

# Display results
print("ASSIGNMENT 2: ITERATIVE POLICY EVALUATION")
print("-" * 48)
print("Number of iterations:", iterations)
print(f"V(S0)        = {V['S0']:.6f}")
print(f"V(S1)        = {V['S1']:.6f}")

print("\nAction Values:")
print(f"Q(S0, Down)  = {Q_S0_Down:.6f}")
print(f"Q(S0, Right) = {Q_S0_Right:.6f}")
print(f"Q(S1, Right) = {Q_S1_Right:.6f}")
print(f"Q(S1, Up)    = {Q_S1_Up:.6f}")

print("\nPolicy Consistency Checks:")
print(f"V(S0) = Q(S0, Down): "
      f"{abs(V['S0'] - Q_S0_Down) < 1e-9}")
print(f"V(S1) = Q(S1, Right): "
      f"{abs(V['S1'] - Q_S1_Right) < 1e-9}")
```
