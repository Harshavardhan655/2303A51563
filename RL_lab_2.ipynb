TASK 1

```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Initial values
N = {'A': 0, 'B': 0, 'C': 0}       # Number of selections
Q = {'A': 0, 'B': 0, 'C': 0}       # Q-values
R = {'A': 0, 'B': 0, 'C': 0}       # Total rewards

epsilon = 0.2
N_play = 10

# Play for 10 times
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Part C - E-Greedy Decision
    if r < epsilon:
        # Exploration
        machine = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        machine = max(Q, key=Q.get)
        decision = "Exploitation"

    # Part D - Reward Collection
    reward = float(input(f"Play {play}: Play with machine {machine}\nEnter reward: "))

    # Part E - Q-Value Update
    N[machine] += 1

    Q[machine] = Q[machine] + (
        reward - Q[machine]
    ) / N[machine]

    R[machine] += reward

    print(f"{decision} -> Machine {machine}")
    print(f"Updated Q({machine}) = {Q[machine]:.2f}\n")


# Part F - Display Results
print("\n========== FINAL RESULTS ==========")

for machine in machines:
    print(f"Machine {machine}")
    print(f"N({machine}) = {N[machine]}")
    print(f"Total Reward = {R[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")
    print()

print("Final Q-values:", Q)
```

```python
import random

# Machines
machines = ['A', 'B', 'C']

# Reward patterns
rewards = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Tables
N = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0, 'B': 0, 'C': 0}

epsilon = 0.2
reward_index = {'A': 0, 'B': 0, 'C': 0}

# 10 plays
for play in range(1, 11):

    r = random.random()

    # Epsilon-greedy selection
    if r < epsilon:
        machine = random.choice(machines)
        decision = "Exploration"
    else:
        machine = max(Q, key=Q.get)
        decision = "Exploitation"

    # Get reward from reward pattern
    index = reward_index[machine]

    # If all rewards are used, start again
    if index >= len(rewards[machine]):
        index = 0

    reward = rewards[machine][index]

    reward_index[machine] += 1

    # Update N
    N[machine] += 1

    # Incremental Q-value update
    Q[machine] = Q[machine] + (
        reward - Q[machine]
    ) / N[machine]

    print(
        f"Play {play}: "
        f"{decision}, "
        f"Machine {machine}, "
        f"Reward {reward}, "
        f"Q = {Q[machine]:.2f}"
    )


# Final results
print("\n========== FINAL RESULTS ==========")

total_reward = 0

for machine in machines:
    print(f"\nMachine {machine}")
    print(f"N({machine}) = {N[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")

print("\nFinal Q Table:")
print(Q)

print("\nTotal reward received after 10 plays:")
print("Calculate by adding all rewards printed above.")
```
